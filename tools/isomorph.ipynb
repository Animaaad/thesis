```python
from sage.all import *
from multiprocessing import Pool
import itertools
from sage.graphs.graph_coloring import edge_coloring

def process_graph(adj):
    G = Graph(adj)
    return G.canonical_label(algorithm="bliss").graph6_string()


def graph_stream(filename):
    with open(filename, "r") as f:
        total_graphs = int(f.readline().strip())

        for _ in range(total_graphs):
            f.readline()  # skip graph ID
            n = int(f.readline().strip())

            adj = {}
            for v in range(n):
                adj[v] = list(map(int, f.readline().split()))
            
            yield adj

def is_cubic(G):
    ok = True
    for v in G.vertices():
        if G.degree(v) != 3:
            ok = False
    return ok
    
def print_graph_adj(G):
    with open("out.txt", "w") as out:
        for v in G.vertices():
            out.write(f"{v}: {G.neighbors(v)}" + "\n")

def deduplicate_parallel(input_file, output_file, processes=4, chunk_size=1):

    seen = set()
    unique_count = 0
    total_processed = 0

    stream = graph_stream(input_file)   # ← create ONCE
    

    with Pool(processes=processes) as pool, open(output_file, "w") as out:

        while True:
            batch = list(itertools.islice(stream, chunk_size))
            if not batch:
                break
            

            results = pool.map(process_graph, batch)

            for g6 in results:
                total_processed += 1
                if(not is_cubic(Graph(g6)) or (edge_coloring(Graph(g6), value_only=True, solver='GLPK') == 3)):
                   print("LALA")

                if g6 not in seen:
                    seen.add(g6)
                    out.write(g6 + "\n")
                    unique_count += 1
                    


            print(f"Processed {total_processed} | Unique: {unique_count}")

    print("Done.")
    print("Total processed:", total_processed)
    print("Unique graphs:", unique_count)


if __name__ == "__main__":
    deduplicate_parallel("snarks.ba", "snarks_unique.g6", processes=4)
```
